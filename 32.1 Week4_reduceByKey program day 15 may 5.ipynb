```python
from pyspark.sql import SparkSession
import getpass
username = getpass.getuser()
spark = SparkSession. \
builder. \
config('spark.ui.port', '0'). \
config("spark.sql.warehouse.dir", f"/user/{username}/warehouse"). \
enableHiveSupport(). \
master('yarn'). \
getOrCreate()
```

## let's try to process some bigger file which is around the 3.5 gb here 

```python
# creating  a base redd here .
base_rdd = spark.sparkContext.textFile("/public/trendytech/orders/orders.csv")
```

```python
# here we need the last column 
mapped_rdd = base_rdd.map(lambda x:(x.split(",")[3],1))
```

# the output will look  like this here 

```python
mapped_rdd.take(5)
```

```python
reduced_rdd = mapped_rdd.reduceByKey(lambda x,y: x+y)
```

```python
reduced_rdd.collect()
```
